```python
class calculator:
 def __init__(self,num1,num2):
  self.num1=num1
  self.num2=num2

 def add(self):
  return self.num1+self.num2
 def sub(self):
  return self.num1-self.num2





cal=calculator(15,5)
print(cal.add())
print(cal.sub())
c=calculator(2000,460)
print(c.add())
print(c.sub())
```

```python
import qrcode
from  IPython.display import display
qr=qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_L,
    box_size=10,
    border=4,
)
qr.add_data=("thi is my first QR  by python!!")
qr.make(fit=True)
img=qr.make_image(fill_color="black",back_color="white")
display(img)
```

```python
#this  is QR code for restorants and big hotells for sample
import qrcode
#IN COLAB not directly visualize  qr code
from IPython.display import display

data="""
   _-_-_-_-_-
    today menu
    _-_-_-_-_-
  FOODlist   price
    dor_wet: 250 ETB
    injera:  40  ETB
    rice:    50  ETB
    beans:   20  ETB
    milk:    20  ETB
    meat:    200 ETB
    egg:     10  ETB
    _-_-_-_-_-_-
    INJOY YOUR  MEAL!!
    -_-_-_-_-_-_-
"""



qr=qrcode.QRCode(
    version=5,
    error_correction=qrcode.constants.ERROR_CORRECT_L,
    box_size=4,
    border=2,

)
qr.add_data(data)
qr.make(fit=True)
img=qr.make_image(fill_color="black",back_color="white")
display(img)


```
